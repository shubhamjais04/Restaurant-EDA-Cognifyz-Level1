# Task 1.1: Data Exploration and Preprocessing

```python
# importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

print("libraries imported successfully!")
```

```python
# load dataset
df = pd.read_csv('Data/Dataset .csv')
df.head()
```

```python
# checking rows and columns
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# checking data types
df.info()
```

```python
# checkig missing values
df.isnull().sum()
```

```python
# handling missing values in Cuisines column
df['Cuisines'] = df['Cuisines'].fillna('Unknown')
df.isnull().sum()
```

### Convert Yes/No Columns to Boolean

```python
df['Has Table booking'] = df['Has Table booking'].map({'Yes': True, 'No': False})
df['Has Online delivery'] = df['Has Online delivery'].map({'Yes': True, 'No': False})
df['Is delivering now'] = df['Is delivering now'].map({'Yes': True, 'No': False})
df['Switch to order menu'] = df['Switch to order menu'].map({'Yes': True, 'No': False})

df.head()
```

```python
# checking Aggregate rating distribution
df['Aggregate rating'].describe()
```

### Plot Rating Distribution

```python
plt.figure(figsize=(10,5))
plt.hist(df['Aggregate rating'], bins=20, edgecolor='black')
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.savefig('../Level_1/Images/rating_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# checking Class imbalance
df['Rating text'].value_counts()
```

```python
# saving cleaned dataset
df.to_csv('Data/cleaned_restaurant_data.csv', index=False)
print("Cleaned dataset saved!")
```

### Task 1.1 Summary

- Loaded dataset with 9551 rows and 21 columns  
- Found missing values only in Cuisines column  
- Filled missing cuisines with "Unknown"  
- Converted Yes/No columns to boolean  
- Ratings range from 0 to 4.9  
- 2148 restaurants have rating 0 (Not rated)  
- Saved cleaned dataset

# Task 1.3: Geospatial Analysis

```python
# load cleaned dataset
df = pd.read_csv('Data/cleaned_restaurant_data.csv')
print("Loaded successfully!")
print("Shape:", df.shape)
df.head()
```

```python
# check co-ordinates
zero_lat = (df['Latitude'] == 0).sum()
zero_lon = (df['Longitude'] == 0).sum()

print(f"Rows with Latitude = 0: {zero_lat}")
print(f"Rows with Longitude = 0: {zero_lon}")
```

```python
# remove invalid coordinates
df_map = df[(df['Latitude'] != 0) & (df['Longitude'] != 0)]
print(f"Rows with valid coordinates: {len(df_map)}")
print(f"Rows removed: {len(df) - len(df_map)}")
```

### Plot Restaurants on Map

```python
# Create scatter plot of restaurant locations
plt.figure(figsize=(12, 8))
plt.scatter(df_map['Longitude'], df_map['Latitude'], alpha=0.5, s=10, c='blue')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Restaurant Locations')
plt.savefig('../Level_1/Images/restaurant_locations.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Restaurants by Country

```python
# Count restaurants by country
country_counts = df['Country Code'].value_counts()
print("Restaurants by Country Code:")
print(country_counts)
```

```python
# Bar plot for countries
plt.figure(figsize=(8,5))
country_counts.plot(kind='bar', color='coral')
plt.title('Number of Restaurants by Country')
plt.xlabel('Country Code')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.savefig('../Level_1/Images/restaurants_by_country.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Restaurants by City (Top 15)

```python
# Top cities
top_cities = df['City'].value_counts().head(15)
print("Top 15 Cities:")
print(top_cities)
```

```python
# Bar plot for cities
plt.figure(figsize=(12,6))
top_cities.plot(kind='bar', color='lightblue')
plt.title('Top 15 Cities with Most Restaurants')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.savefig('../Level_1/Images/top_cities_map.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Average Rating by City (Top 15)

```python
# Calculate average rating per city
city_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(15)
print("Top 15 Cities by Average Rating:")
print(city_rating)
```

```python
# Bar plot for average rating by city
plt.figure(figsize=(12,6))
city_rating.plot(kind='bar', color='green')
plt.title('Top 15 Cities by Average Rating')
plt.xlabel('City')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.savefig('../Level_1/Images/avg_rating_by_city.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Average Rating by Country

```python
# Calculate average rating per country
country_rating = df.groupby('Country Code')['Aggregate rating'].mean().sort_values(ascending=False)
print("Average Rating by Country Code:")
print(country_rating)
```

```python
# Bar plot for average rating by country
plt.figure(figsize=(8,5))
country_rating.plot(kind='bar', color='purple')
plt.title('Average Rating by Country')
plt.xlabel('Country Code')
plt.ylabel('Average Rating')
plt.xticks(rotation=0)
plt.savefig('../Level_1/Images/avg_rating_by_country.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Rating vs Location Analysis

```python
# Scatter plot with rating as color
plt.figure(figsize=(12, 8))
scatter = plt.scatter(df_map['Longitude'], df_map['Latitude'], 
                      c=df_map['Aggregate rating'], cmap='viridis', 
                      alpha=0.6, s=15)
plt.colorbar(scatter, label='Rating')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Restaurant Locations Colored by Rating')
plt.savefig('../Level_1/Images/rating_vs_location.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Conclusion - Correlation between Location and Rating

```python
# Check if there's a pattern
print("Observations:")
print(f"- Highest rated cities: {city_rating.head(3).index.tolist()}")
print(f"- Highest rated country: {country_rating.head(1).index[0]} with rating {country_rating.head(1).values[0]:.2f}")
print("\nConclusion: Location does affect ratings. Different cities and countries have different average ratings.")
```
